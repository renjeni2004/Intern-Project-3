This notebook is an exploratory entry point. The reproducible training pipeline is implemented in `src/train.py`. The included dataset is synthetic.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
ROOT = Path.cwd()
if not (ROOT / 'data/raw/customer_churn_synthetic.csv').exists():
    ROOT = ROOT.parent
DATA = ROOT / 'data/raw/customer_churn_synthetic.csv'
df = pd.read_csv(DATA)
df.head()
```

## 1. Dataset inspection

```python
print('Shape:', df.shape)
display(df.info())
display(df.isna().sum().sort_values(ascending=False).head(10))
display(df.describe(include='all').T.head(20))
```

## 2. Target distribution

```python
ax = df['Churn'].value_counts(normalize=True).mul(100).plot(kind='bar', title='Churn distribution')
ax.set_ylabel('Share (%)'); plt.tight_layout(); plt.show()
```

## 3. Numeric relationships

```python
numeric = df.select_dtypes(include='number')
fig, ax = plt.subplots(figsize=(8,5))
import seaborn as sns
sns.heatmap(numeric.corr(), cmap='coolwarm', center=0, ax=ax)
ax.set_title('Numeric feature correlation'); plt.tight_layout(); plt.show()
```
